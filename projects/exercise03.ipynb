## Polynomial Interpolation

Polynomial interpolation is a process of finding a polynomial function that precisely passes through a given set of points. This technique is fundamental in various areas of numerical analysis, science, and engineering for approximating functions.

In this lesson, we will explore how to use linear algebra concepts, specifically matrices and vectors, to perform polynomial interpolation. The hope is that this will not only reinforce your understanding of linear algebra but also introduce you to related practical applications.

#### Conceptual Overview

The idea behind polynomial interpolation is straightforward: given a set of $n+1$ distinct points $(x_0,y_0),(x_1,y_1) \cdots (x_n,y_n)$, we aim to find a polynomial of degree $n$ that passes through all these points. The polynomial is typically represented as:

$$P(x) = a_0 + a_1x + a_2x^2 + \cdots + a_nx^n$$

where $a_0, a_1, \ldots, a_n$ are the coefficients we need to determine. Plugging in all $n+1$ points yields the set of equations
$$
\begin{aligned}
    a_0 + a_1x_0 + a_2x_0^2 + \cdots + a_n{x_0}^n &= y_0 \\
    a_0 + a_1x_1 + a_2x_1^2 + \cdots + a_n{x_1}^n &= y_1 \\
    \vdots \\
    a_0 + a_1x_n + a_2x_n^2 + \cdots + a_n{x_n}^n &= y_n \\
\end{aligned}
$$
which can be rewritten as
$$
    \begin{bmatrix}
    1 & x_0 & x_0^2 & \cdots & x_0^n \\
    1 & x_1 & x_1^2 & \cdots & x_1^n \\
    1 & x_2 & x_2^2 & \cdots & x_2^n \\
    \vdots \\
    1 & x_n & x_n^2 & \cdots & x_n^n
    \end{bmatrix}
    \begin{bmatrix}
    a_0 \\ a_1 \\ a_2 \\ \vdots \\ a_n
    \end{bmatrix}
    =
    \begin{bmatrix}
    y_0 \\ y_1 \\ y_2 \\ \vdots \\ y_n
    \end{bmatrix}
$$

Since we know all of the $(x_i,y_i)$ and our goal is to find the values of $a_i$ then this becomes a simple exercise in matrix reduction.

## Least Squares Linear Regression

In the final step of the previous section, you likely noticed that while our degree 20 polynomial perfectly passed through all 21 points, the resulting graph was wild and erratic. In numerical analysis, this is known as **overfitting**. 

Instead of forcing a complex curve to hit every single point perfectly, what if we just wanted to find a simple straight line that captures the general trend of the data? This is called **linear regression**, and we want to find a "line of best fit" in the form:

$$y = c_0 + c_1x$$

Let's start with a simpler example of 3 points that do *not* form a perfectly straight line: $(1, 1), (2, 3)$, and $(3, 2)$. If we try to find a line that passes perfectly through all three, we get this system of equations:

$$c_0 + c_1(1) = 1$$
$$c_0 + c_1(2) = 3$$
$$c_0 + c_1(3) = 2$$

Rewritten as an augmented matrix $[A | \vec{b}]$, this gives us:

$$ \begin{bmatrix} 1 & 1 & | & 1 \\ 1 & 2 & | & 3 \\ 1 & 3 & | & 2 \end{bmatrix} $$

Let's see what happens when we try to solve this using matrix reduction.

If you look at the last row of the reduced matrix, you should see `[0, 0, 1]`. In equation form, this translates to $0 = 1$. Because the points don't lie on a single line, this system is **inconsistent** (overdetermined) and has no exact solution! 

**Calculus to the Rescue: The "Least Squares" Method**
Since we can't find a perfect line, we want to find the line that generates the *least* amount of error. For any chosen $c_0$ and $c_1$, the error (or residual) at a given point is the difference between our line's prediction and the actual $y$-value:
$$ \text{Error}_i = (c_0 + c_1x_i) - y_i $$

To prevent negative errors from cancelling out positive errors, we square them. We then want to **minimize the Sum of the Squared Errors (SSE)**:
$$ SSE = \sum_{i=1}^{n} \left( (c_0 + c_1x_i) - y_i \right)^2 $$

From calculus, we know that to find the minimum of a function, we take the derivative and set it to zero. Here we have two real variables ($c_0$ and $c_1$), as you will see in a future course a necessary condition for a minimum in this case will require the derivative with respect to each variable (aka the partial derivatives) to both be zero:

1. **Derivative with respect to $c_0$:** 
   $$\sum 2\left(c_0 + c_1x_i - y_i\right)(1) = 0 \implies \sum (c_0 + c_1x_i) = \sum y_i$$
2. **Derivative with respect to $c_1$:** 
   $$\sum 2\left(c_0 + c_1x_i - y_i\right)(x_i) = 0 \implies \sum x_i(c_0 + c_1x_i) = \sum x_i y_i$$
   
This yields two equations in the two unknowns $c_0$ and $c_1$.

**The Matrix Connection**
Here is the magical part: we can generate those exact calculus equations using matrix algebra! Notice that our matrix $A$ consists of a column of $1$s and a column of $x$ values. The transpose of $A$, written as $A^T$, turns those columns into rows. 

If we multiply our original impossible equation $A\overrightarrow{c} = \overrightarrow{b}$ by $A^T$ on both sides, we get:

$$A^T A \overrightarrow{c} = A^T \overrightarrow{b}$$

Let's look at what multiplying by $A^T$ actually does:
$$ \begin{bmatrix} 1 & 1 & 1 \\ x_1 & x_2 & x_3 \end{bmatrix} \begin{bmatrix} c_0 + c_1 x_1 \\ c_0 + c_1 x_2 \\ c_0 + c_1 x_3 \end{bmatrix} = \begin{bmatrix} 1 & 1 & 1 \\ x_1 & x_2 & x_3 \end{bmatrix} \begin{bmatrix} y_1 \\ y_2 \\ y_3 \end{bmatrix} $$

If you perform this multiplication you will see that the top row perfectly recreates the first calculus equation above, and the bottom row perfectly recreates the second equation! 

By multiplying by $A^T$, we transform our impossible system into a solvable system known as the **Normal Equations**. Because $A^T A$ is a square matrix, we can now solve for our coefficients $\vec{c}$ using an inverse matrix:

$$\overrightarrow{c} = (A^T A)^{-1} A^T \overrightarrow{b}$$

**A Quick Disclaimer on Invertibility**
Note that $(A^T A)$ is only invertible if the columns of our original matrix $A$ are **linearly independent**. (You can try to prove this by showing that the only solution to $A^T A \overrightarrow{x} = \overrightarrow{0}$ is the trivial solution.)

In the context of fitting a line $y = c_0 + c_1x$, our matrix $A$ has a column of $1$s and a column of $x$-values. These columns are only linearly dependent if all our $x$-values are exactly the same (which would mean our data is just a vertical line, and a standard function couldn't model it anyway). As long as we have at least two *different* $x$-coordinates in our dataset, the columns of $A$ are independent, $(A^T A)$ is invertible, and the method should work.
